### 1) Ver o que tem de errado

```python
import pandas as pd

df = pd.read_csv('imoveis_sujos.csv')
print(df)
df.info()
print(df.isnull().sum())
print('duplicatas:', df.duplicated().sum())
```

### 2) Bairro

```python
df['bairro'] = df['bairro'].str.lower().str.strip().str.capitalize()
print(df['bairro'].unique())
```

### 3) Quartos

```python
df['quartos'] = df['quartos'].replace('dois', 2)
df['quartos'] = pd.to_numeric(df['quartos'], errors='coerce')
df['quartos'] = df['quartos'].fillna(0).astype(int)
print(df['quartos'])
```

### 4) Área

```python
df['area_m2'] = pd.to_numeric(df['area_m2'], errors='coerce')
mediana_area = df['area_m2'].median()
df['area_m2'] = df['area_m2'].fillna(mediana_area)
print('mediana area:', mediana_area)
print(df[['id', 'area_m2']])
```

### 5) Aluguel

```python
df['aluguel'] = (
    df['aluguel'].astype(str)
    .str.replace('R$', '', regex=False)
    .str.strip()
    .str.replace('.', '', regex=False)
    .str.replace(',', '.')
)
df['aluguel'] = pd.to_numeric(df['aluguel'])
print(df['aluguel'])
```

### 6) Condomínio

```python
mediana_cond = df['condominio'].median()
df['condominio'] = df['condominio'].fillna(mediana_cond)
print('mediana condominio:', mediana_cond)
print('nulos:', df['condominio'].isnull().sum())
```

### 7) Duplicatas

```python
print('antes:', df.duplicated().sum())
df = df.drop_duplicates().reset_index(drop=True)
print('depois:', df.duplicated().sum())
print(df)
```

### 8) Conferir e salvar

```python
print(df.info())
print(df.describe())
df.to_csv('imoveis_limpos.csv', index=False)
print('salvo imoveis_limpos.csv')
df
```
