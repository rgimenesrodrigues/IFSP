## Passo 1 — Diagnóstico

```python
import pandas as pd

df = pd.read_csv('imoveis_sujos.csv')
print(df)
print()
df.info()
print()
print('Nulos por coluna:')
print(df.isnull().sum())
print()
print('Duplicatas:', df.duplicated().sum())
```

Problemas encontrados:
1. Bairros com caixa diferente e espaços (`centro `, `jardins`, `vila mariana`)
2. `quartos` com texto (`dois`)
3. `area_m2` com NaN
4. `aluguel` em formato de moeda brasileira
5. `condominio` com NaN
6. Linhas duplicadas depois da padronização

## Passo 2 — Limpar coluna `bairro`

```python
df['bairro'] = df['bairro'].str.lower().str.strip().str.capitalize()
print(df['bairro'].unique())
```

## Passo 3 — Corrigir texto em coluna numérica (`quartos`)

```python
df['quartos'] = df['quartos'].replace('dois', 2)
df['quartos'] = pd.to_numeric(df['quartos'], errors='coerce')
df['quartos'] = df['quartos'].fillna(0).astype(int)
print(df['quartos'])
```

## Passo 4 — Converter `area_m2` e imputar mediana

```python
df['area_m2'] = pd.to_numeric(df['area_m2'], errors='coerce')
mediana_area = df['area_m2'].median()
print('Mediana da área:', mediana_area)
df['area_m2'] = df['area_m2'].fillna(mediana_area)
print(df[['id', 'area_m2']])
```

## Passo 5 — Limpar moeda (`aluguel`)

```python
df['aluguel'] = (
    df['aluguel']
    .astype(str)
    .str.replace('R$', '', regex=False)
    .str.strip()
    .str.replace('.', '', regex=False)
    .str.replace(',', '.')
)
df['aluguel'] = pd.to_numeric(df['aluguel'])
print(df['aluguel'])
print('Soma dos aluguéis:', df['aluguel'].sum())
```

## Passo 6 — Preencher `condominio` com a mediana

```python
mediana_cond = df['condominio'].median()
print('Mediana do condomínio:', mediana_cond)
df['condominio'] = df['condominio'].fillna(mediana_cond)
print('Nulos em condominio:', df['condominio'].isnull().sum())
```

## Passo 7 — Remover duplicatas

```python
print('Duplicatas antes:', df.duplicated().sum())
df = df.drop_duplicates().reset_index(drop=True)
print('Duplicatas depois:', df.duplicated().sum())
print('Linhas finais:', len(df))
```

## Passo 8 — Conferência final

```python
print(df.info())
print()
print(df.describe())
print()
df
```

## Passo 9 — Exportar CSV limpo

```python
df.to_csv('imoveis_limpos.csv', index=False)
print('Dataset limpo salvo em imoveis_limpos.csv')
```

## Resumo das decisões
- Bairro padronizado com `lower`, `strip` e `capitalize`
- Texto `dois` virando `2` nos quartos
- Área e condomínio preenchidos com **mediana** (menos sensível a outlier do que a média)
- Aluguel convertido de `R$ 1.500,00` para float
- Duplicatas removidas no final
