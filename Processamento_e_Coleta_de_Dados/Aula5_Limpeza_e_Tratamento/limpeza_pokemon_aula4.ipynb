# Tarefa de casa — Limpar dataset da Aula 4
Aplicar o pipeline de limpeza no CSV de Pokémon coletado na aula anterior.


```python
import pandas as pd
from pathlib import Path

origem = Path('../Aula4_Projeto_Coleta/pokemon_dataset.csv')
df = pd.read_csv(origem)
print('Antes:')
print(df.shape)
print(df.dtypes)
print(df.isnull().sum())
df.head()
```

```python
df_limpo = df.copy()

# padronizar texto
df_limpo['nome'] = df_limpo['nome'].astype(str).str.strip().str.title()
df_limpo['tipo_principal'] = df_limpo['tipo_principal'].astype(str).str.strip().str.lower()
df_limpo['tipo_secundario'] = df_limpo['tipo_secundario'].astype(str).str.strip().str.lower()
df_limpo['tipo_secundario'] = df_limpo['tipo_secundario'].replace({'nan': 'nenhum', '': 'nenhum'})

# garantir tipos numéricos
cols_num = ['altura_m', 'peso_kg', 'experiencia_base', 'hp', 'ataque', 'defesa',
            'ataque_especial', 'defesa_especial', 'velocidade', 'total_stats',
            'qtd_habilidades', 'qtd_movimentos']
for col in cols_num:
    df_limpo[col] = pd.to_numeric(df_limpo[col], errors='coerce')

# recalcular total_stats por segurança
df_limpo['total_stats'] = (
    df_limpo['hp'] + df_limpo['ataque'] + df_limpo['defesa']
    + df_limpo['ataque_especial'] + df_limpo['defesa_especial'] + df_limpo['velocidade']
)

# remover duplicatas e linhas sem id/nome
df_limpo = df_limpo.drop_duplicates(subset=['id'])
df_limpo = df_limpo.dropna(subset=['id', 'nome'])
df_limpo = df_limpo.sort_values('id').reset_index(drop=True)

print('Depois:')
print(df_limpo.shape)
print(df_limpo.isnull().sum())
df_limpo.head()
```

```python
df_limpo.to_csv('pokemon_limpo.csv', index=False)
print('Salvo: pokemon_limpo.csv')
```
